```python
# Step 2: Setup Azure OpenAI client
import os
from dotenv import load_dotenv
from openai import AzureOpenAI

# Load .env file
load_dotenv()

# Read Azure credentials
AZURE_OPENAI_ENDPOINT = os.getenv("AZURE_OPENAI_ENDPOINT")
AZURE_OPENAI_API_KEY = os.getenv("AZURE_OPENAI_API_KEY")
AZURE_OPENAI_API_VERSION = os.getenv("AZURE_OPENAI_API_VERSION")
AZURE_OPENAI_DEPLOYMENT_NAME = os.getenv("AZURE_OPENAI_DEPLOYMENT_NAME")

# Initialize Azure OpenAI client
client = AzureOpenAI(
    api_key=AZURE_OPENAI_API_KEY,
    api_version=AZURE_OPENAI_API_VERSION,
    azure_endpoint=AZURE_OPENAI_ENDPOINT
)

print("✅ Azure OpenAI setup complete")
```

```python
# Step 3: Paste your Scope of Work text here
document_text = """
• Provide a summary of the scope of work specific to your engineering discipline.
 Background situation overview and site conditions:
 Merivale TS is one of two major supply stations located in Ottawa. The station 
comprises of 230kV Gas Insulated Switchgear (GIS), two 230/115kV autotransformers 
and 115kV switchyard. The station supplies 115kV load in the Greater Ottawa region.
 Multiple assets at Merivale TS including six (6) 230kV GIS breakers, one (1) 
230/115kV autotransformer and four (4) 115kV oil circuit breakers are approaching 
end of life. The replacement of these assets is prudent to maintain reliability of 
supply in the region. 
 This project will also address the replacement of disconnect switches, instrument 
transformers and cap & pin strain insulators that become redundant or have reached 
end of life. The protection, control and telecom facilities and AC station service 
supply will also be upgraded consistent with present Hydro One design requirements 
and standards. 
 Mechanical SOW:
1. Provide arrangement layout of HADs for new T24 transformer (design will be done 
under AR 23709).
2. Review existing HVAC and FAS* systems in the existing buildings on the site, where 
electrical equipment will be replaced. 
*Note: by electrical engineer
"""

# Build the prompt
prompt = f"""
You are an expert business and technical writer. Your task is to rewrite the given 'Scope of Work' text into a clear, professional description.

Rules:
- Do not miss or remove any technical keywords.
- Do not alter the meaning or context of the scope of work.
- If the scope of work is long (more than ~150 words), compress it into a concise description, ideally around 100–200 words.
- If the scope of work is short (1–2 lines), keep it concise and do not artificially expand it.
- Output only a natural description in plain text (no bullet points, no lists, no headings).

Scope of Work text:
{document_text}

Now provide the rewritten description:
"""
```

```python
# Step 4: Call Azure OpenAI GPT-4o with the prompt
response = client.chat.completions.create(
    model=AZURE_OPENAI_DEPLOYMENT_NAME,  # e.g. "gpt-4o"
    messages=[
        {"role": "system", "content": "You are a helpful assistant for professional summarization."},
        {"role": "user", "content": prompt}
    ],
    temperature=0.3,
    max_tokens=500
)

# Print the description
print("Scope of Work Description:\n")
print(response.choices[0].message.content)
```
